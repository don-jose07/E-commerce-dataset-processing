import librarys


```python
import pandas as pd
import numpy as np
```

```python

customers=pd.read_csv("Customers.csv")
orders=pd.read_csv("Orders.csv")
products=pd.read_csv("Products.csv")
```

```python
# Merge orders with customers
orders_customers = pd.merge(orders, customers, on='Customer_ID', how='left')

# Merge the combined orders_customers with products
processed_df = pd.merge(orders_customers, products, on='Product_ID', how='left')

print("Merged DataFrame Head:")
print(processed_df.head())
print("\nMerged DataFrame Info:")
processed_df.info()
```

```python
# Convert 'Order_Date' to datetime objects
processed_df['Order_Date'] = pd.to_datetime(processed_df['Order_Date'])

# Extract DateTime features
processed_df['Order_Month'] = processed_df['Order_Date'].dt.month
processed_df['Order_Day'] = processed_df['Order_Date'].dt.day
processed_df['Order_DayOfWeek'] = processed_df['Order_Date'].dt.dayofweek # Monday=0, Sunday=6
processed_df['Order_Hour'] = processed_df['Order_Date'].dt.hour
processed_df['Order_Year'] = processed_df['Order_Date'].dt.year

print("DataFrame with new Date and Time features:")
print(processed_df[['Order_Date', 'Order_Month', 'Order_Day', 'Order_DayOfWeek', 'Order_Hour', 'Order_Year']].head())
processed_df.info()
```

```python
# Create a 'Total_Price' column using apply()
processed_df['Total_Price'] = processed_df.apply(lambda row: row['Quantity'] * row['Unit_Price'], axis=1)

print("DataFrame with 'Total_Price' column:")
print(processed_df[['Quantity', 'Unit_Price', 'Total_Price']].head())

# Demonstrate concat() by adding a dummy DataFrame
dummy_data = {
    'Order_ID': ['O121', 'O122'],
    'Customer_Name': ['Test User 1', 'Test User 2'],
    'Total_Price': [100, 200]
}
dummy_df = pd.DataFrame(dummy_data)

# For demonstration, only concatenate relevant columns from processed_df
# Note: In a real scenario, you'd ensure columns match for meaningful concatenation
# We'll select a few common columns for this demonstration.

# To demonstrate concat() meaningfully, let's create a small temporary dataframe
# with some of the columns already present in `processed_df` and some new ones.
# This is purely for demonstrating `concat()` as per the prompt, not for actual data processing.

print("\nDemonstrating concat() with a dummy DataFrame:")
# Create a simple DataFrame for demonstration of concat
demo_df1 = processed_df[['Order_ID', 'Customer_ID', 'Order_Date', 'Total_Price']].head(3)
demo_df2 = pd.DataFrame({
    'Order_ID': ['D001', 'D002'],
    'Customer_ID': ['C031', 'C032'],
    'Order_Date': pd.to_datetime(['2026-04-01', '2026-04-02']),
    'Total_Price': [500, 750]
})

# Concatenate the two demonstration DataFrames
concatenated_df_demo = pd.concat([demo_df1, demo_df2], ignore_index=True)
print(concatenated_df_demo)
```

```python
# Export the final processed dataset to a CSV file
output_filename = 'processed_ecommerce_dataset.csv'
processed_df.to_csv(output_filename, index=False)

print(f"Final processed dataset exported to '{output_filename}'")
print("Head of the exported DataFrame:")
print(processed_df.head())
```
